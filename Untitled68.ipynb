```python
bo  = ["Customer_Name","Order_Time","Region","Sales_Representative","Shipping_Method","Delivery_Days"]
chu = ["Customer_Gender","Customer_Segment","Sales_Channel","Payment_Method","Order_Status","Return_Reason","City"]

df = df.drop(columns=bo)
df[chu] = df[chu].apply(lambda s: s.str.strip().str.title())
df["Promotion_Code"] = df.Promotion_Code.str.strip().str.upper()
df["Product_Category"] = df["Product_Category"].str.strip().str.title()
```

```python
result = duckdb.query("""
    SELECT
        Order_Year,
        COUNT(DISTINCT Order_ID ) AS total_orders,
        SUM(Profit) AS total_profit
    FROM df
    WHERE Order_Year IS NOT NULL
    GROUP BY Order_Year
    ORDER BY Order_Year
""").to_df()

print(result)
```

```python
df_dax = duckdb.query("""
    SELECT
Order_Year,
Sales_Channel,
Product_Category,
Quantity,
Unit_Price,
Sales_Amount,
Cost_Amount,
Profit,
Order_ID,
FROM df
""").to_df()
```

```python
margin_by_year = duckdb.query("""
SELECT
  Order_Year,
  SUM(Sales_Amount) AS total_sales,
  SUM(Cost_Amount) AS total_cost,
  SUM(Profit) AS total_profit,
ROUND(SUM(Profit) / NULLIF(SUM(Cost_Amount), 0), 4) AS margin_ratio
 FROM df
 WHERE Order_Year IS NOT NULL
GROUP BY Order_Year
 ORDER BY Order_Year
""").df()
print(margin_by_year.to_string())
```
